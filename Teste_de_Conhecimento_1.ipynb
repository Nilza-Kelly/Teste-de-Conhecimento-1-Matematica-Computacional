```python
# Teste de Conhecimento 1

## Análise e previsão dos custos de transporte

## Este notebook apresenta uma análise dos custos mensais de transporte de uma empresa de comércio eletrônico.
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter
import numpy as np
```

```python
## Etapa 1 – Carregar os dados

dados = pd.read_csv("base_custos_transporte.csv")

dados
```

```python
## Etapa 2 – Identificação das variáveis

### a) Qual é a variável independente?
# A variável independente é o mês

### b) Qual é a variável dependente?
# A variável dependente é o custo de transporte, porque seu comportamento está sendo analisado em função do mês.

### c) O que representa cada variável?
# A variável "mes" representa o período de tempo analisado, variando do mês 1 ao mês 12. A variável "custo_transporte" representa o custo total de transporte da empresa em cada mês, medido em reais.
```

```python
## Etapa 3 - Construir um gráfico

plt.figure(figsize=(10,6))

plt.scatter(
    dados["mes"],
    dados["custo_transporte"]
)

plt.title("Gráfico - Custo de transporte por mês")
plt.xlabel("Mês")
plt.ylabel("Custo de transporte em R$")

plt.grid(True)

plt.show()

### Análise do gráfico
# Observando o gráfico, é possível perceber que os custos de transporte apresentam uma tendência de crescimento ao longo dos meses.
# Apesar de existirem pequenas variações entre alguns meses, de forma geral o custo aumenta conforme o mês aumenta.
# No primeiro mês, o custo é de R$ 44.500,00, enquanto no décimo segundo mês chega a R$ 64.000,00. Portanto, os dados apresentam uma tendência crescente.
```

```python
## Etapa 4 - Encontrar uma função

# x = mês / y = custo de transporte

x = dados["mes"]
y = dados["custo_transporte"]

a, b = np.polyfit(x, y, 1)

print("Coeficiente angular(a):", a)
print("Coeficiente linear(b):", b)
print(f"Função encontrada: f(x) = {a: .4f}x + {b: .2f}")
```

```python
### Cálculo da função

n = len(x)

soma_x = x.sum()
soma_y = y.sum()
soma_xy = (x * y).sum()
soma_x2 = (x ** 2).sum()

a_calculado = (
    n * soma_xy - soma_x * soma_y
) / (
    n * soma_x2 - soma_x ** 2
)

b_calculado = (
    soma_y - a_calculado * soma_x
) / n

print("n =", n)
print("Σx =", soma_x)
print("Σy =", soma_y)
print("Σxy =", soma_xy)
print("Σx² =", soma_x2)

print("\na =", a_calculado)
print("b =", b_calculado)

print(f"\nFunção: f(x) = {a_calculado:.2f}x + {b_calculado:.2f}")
```

```python
## Etapa 5 - Interpretar a função

### A função encontrada foi:
# f(x) = 1791,61x + 42204,55

### a) O que representa o coeficiente a?
# O coeficiente "a", representa a variação média estimada do custo de transporte a cada mês.

### b) O que representa o coeficiente b?
# O coeficiente "b", representa o valor em que a reta interceptaria o eixo dos custos quando "x = 0".

### c) O valor de a indica crescimento ou redução dos custos?
#O valor de "a" é positivo, então, a função indica uma tendência de crescimento dos custos de transporte ao longo dos meses.
```

```python
## Etapa 6 - Representar a função graficamente

x_reta = np.linspace(1, 12, 100)

y_reta = a * x_reta + b

plt.figure(figsize=(10, 6))

plt.scatter(
    x,
    y,
    label="Dados reais"
)

plt.plot(
    x_reta,
    y_reta,
    label="Função de tendência"
)

plt.title("Custos de transporte e função de tendência")
plt.xlabel("Mês")
plt.ylabel("Custo de transporte em R$")

plt.legend()
plt.grid(True)

plt.show()

### Análise do gráfico
# O gráfico apresenta os valores dos custos e a reta obtida pela função afim.
# A reta possui inclinação positiva representando de forma aproximada o comportamento geral dos custos, permitindo utilizá-la para realizar estimativas futuras.
```

```python
## Etapa 7 - Fazer previsões

meses_futuros = [13, 14, 15]

for mes in meses_futuros:
    previsao = a * mes + b
    print(f"f({mes}) = {a:.2f} × {mes} + {b:.2f}")
    print(f"Mês {mes}: R$ {previsao:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."))
```

```python
## Etapa 8 – Análise dos resultados

### a) A função encontrada representa bem a tendência dos dados?
# Sim. A função representa de forma aproximada a tendência geral dos dados, pois os custos apresentam um comportamento predominantemente crescente.

### b) Os custos estão aumentando ou diminuindo?
# Os custos estão aumentando. Isso pode ser observado tanto nos valores da tabela quanto na inclinação positiva da reta encontrada.

### c) Aproximadamente quanto o custo aumenta a cada mês?
# De acordo com o coeficiente angular da função, o custo aumenta aproximadamente **R$ 1.791,61 por mês**.
```

```python
## Etapa 9 – Conclusão

# A análise dos dados mostra que os custos de transporte da empresa apresentam uma tendência de crescimento ao longo dos 12 meses observados.
# O coeficiente angular positivo indica que o custo de transporte aumenta, em média, aproximadamente R$ 1.791,61 por mês.
# Portanto, caso o comportamento observado continue semelhante, a empresa pode esperar que seus custos de transporte continuem aumentando nos próximos meses.
# A função encontrada pode ser utilizada como uma estimativa da tendência, mas os valores reais podem apresentar variações em relação às previsões.
```
